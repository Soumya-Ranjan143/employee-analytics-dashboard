```python
import sqlite3
import pandas as pd
import numpy as np

# -------------------------------------------------------------
# 1. DATABASE SETUP & INITIALIZATION
# -------------------------------------------------------------
conn = sqlite3.connect('employee_analytics.db')
cursor = conn.cursor()

# Enable foreign key support in SQLite
cursor.execute("PRAGMA foreign_keys = ON;")

# Create tables
cursor.execute('''
CREATE TABLE IF NOT EXISTS departments (
    dept_id INTEGER PRIMARY KEY,
    dept_name TEXT NOT NULL,
    budget INTEGER
);
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS employees (
    emp_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    dept_id INTEGER,
    salary INTEGER,
    city TEXT,
    join_date TEXT,
    FOREIGN KEY (dept_id) REFERENCES departments(dept_id)
);
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS projects (
    project_id INTEGER PRIMARY KEY,
    project_name TEXT NOT NULL,
    emp_id INTEGER,
    status TEXT,
    FOREIGN KEY (emp_id) REFERENCES employees(emp_id)
);
''')

# Clear old records if re-running
cursor.execute("DELETE FROM projects;")
cursor.execute("DELETE FROM employees;")
cursor.execute("DELETE FROM departments;")

# Populate Department Data
departments_data = [
    (1, 'Engineering', 5500000),
    (2, 'Data Science', 4800000),
    (3, 'Marketing', 2200000),
    (4, 'Human Resources', 1500000),
    (5, 'Sales', 3200000)
]
cursor.executemany("INSERT INTO departments VALUES (?, ?, ?);", departments_data)

# Populate Employee Data (15 records)
employees_data = [
    (101, 'Aarav Sharma', 1, 95000, 'Mumbai', '2022-01-15'),
    (102, 'Siddharth Rao', 2, 120000, 'Bangalore', '2021-06-10'),
    (103, 'Ananya Iyer', 2, 85000, 'Chennai', '2023-02-01'),
    (104, 'Priya Patel', 3, 45000, 'Mumbai', '2022-11-20'),
    (105, 'Suresh Kumar', 1, 60000, 'Delhi', '2020-03-12'),
    (106, 'Amit Verma', 1, 135000, 'Bangalore', '2019-08-05'),
    (107, 'Sneha Reddy', 4, 38000, 'Hyderabad', '2023-05-18'),
    (108, 'Rohan Gupta', 5, 52000, 'Pune', '2021-09-30'),
    (109, 'Sunita Joshi', 3, 48000, 'Delhi', '2022-04-14'),
    (110, 'Vikram Singh', 2, 125000, 'Bangalore', '2020-01-10'),
    (111, 'Neha Kapoor', 4, 42000, 'Mumbai', '2023-01-08'),
    (112, 'Aditya Mehta', 5, 72000, 'Pune', '2021-12-01'),
    (113, 'Shweta Dash', 1, 88000, 'Hyderabad', '2022-07-22'),
    (114, 'Karan Nair', 2, 98000, 'Chennai', '2021-04-19'),
    (115, 'Arjun Saxena', 5, 55000, 'Mumbai', '2023-03-11')
]
cursor.executemany("INSERT INTO employees VALUES (?, ?, ?, ?, ?, ?);", employees_data)

# Populate Project Data (10 records)
projects_data = [
    (1, 'AI Voice Assistant', 102, 'In Progress'),
    (2, 'E-Commerce Portal', 101, 'Completed'),
    (3, 'HR Onboarding Tool', 107, 'Completed'),
    (4, 'Ad Campaign 2024', 104, 'In Progress'),
    (5, 'ETL Data Pipeline', 110, 'In Progress'),
    (6, 'Mobile App Redesign', 106, 'Not Started'),
    (7, 'SEO Optimization', 109, 'Completed'),
    (8, 'CRM Integration', 112, 'In Progress'),
    (9, 'Cloud Migration', 113, 'Not Started'),
    (10, 'Recommendation System', 103, 'Completed')
]
cursor.executemany("INSERT INTO projects VALUES (?, ?, ?, ?);", projects_data)

conn.commit()
print("Database initialized and populated successfully.")

# -------------------------------------------------------------
# 2. SQL QUERIES (pd.read_sql)
# -------------------------------------------------------------

# Q1: High to low salaries with city
q1 = "SELECT name, salary, city FROM employees ORDER BY salary DESC;"
df_q1 = pd.read_sql_query(q1, conn)
print("\n--- Q1: Employees by Highest Salary ---")
print(df_q1.head())

# Q2: Mumbai or Bangalore earners above 50k
q2 = """
SELECT * FROM employees 
WHERE city IN ('Mumbai', 'Bangalore') AND salary > 50000;
"""
df_q2 = pd.read_sql_query(q2, conn)

# Q3: Names starting with A or S
q3 = "SELECT * FROM employees WHERE name LIKE 'A%' OR name LIKE 'S%';"
df_q3 = pd.read_sql_query(q3, conn)

# Q4: Employee name + Dept name
q4 = """
SELECT e.name, d.dept_name 
FROM employees e 
JOIN departments d ON e.dept_id = d.dept_id;
"""
df_q4 = pd.read_sql_query(q4, conn)

# Q5: All departments count (including 0 employees)
q5 = """
SELECT d.dept_name, COUNT(e.emp_id) AS total_employees 
FROM departments d 
LEFT JOIN employees e ON d.dept_id = e.dept_id 
GROUP BY d.dept_id, d.dept_name;
"""
df_q5 = pd.read_sql_query(q5, conn)

# Q6: Employee + Dept + Assigned Project
q6 = """
SELECT e.name, d.dept_name, p.project_name 
FROM employees e 
JOIN departments d ON e.dept_id = d.dept_id 
JOIN projects p ON e.emp_id = p.emp_id;
"""
df_q6 = pd.read_sql_query(q6, conn)

# Q7: Department salary metrics
q7 = """
SELECT 
    d.dept_name, 
    ROUND(AVG(e.salary), 2) AS avg_salary, 
    SUM(e.salary) AS total_payroll, 
    COUNT(e.emp_id) AS total_staff 
FROM departments d 
JOIN employees e ON d.dept_id = e.dept_id 
GROUP BY d.dept_id;
"""
df_q7 = pd.read_sql_query(q7, conn)

# Q8: Depts with avg salary > 60k
q8 = """
SELECT d.dept_name, AVG(e.salary) AS avg_salary 
FROM departments d 
JOIN employees e ON d.dept_id = e.dept_id 
GROUP BY d.dept_id 
HAVING AVG(e.salary) > 60000;
"""
df_q8 = pd.read_sql_query(q8, conn)

# Q9: Earners above company average
q9 = """
SELECT name, salary, city 
FROM employees 
WHERE salary > (SELECT AVG(salary) FROM employees);
"""
df_q9 = pd.read_sql_query(q9, conn)

# Q10: 2nd highest salary
q10 = """
SELECT e.name, d.dept_name, e.salary 
FROM employees e 
JOIN departments d ON e.dept_id = d.dept_id 
ORDER BY e.salary DESC 
LIMIT 1 OFFSET 1;
"""
df_q10 = pd.read_sql_query(q10, conn)

# -------------------------------------------------------------
# 3. PYTHON ANALYSIS (PANDAS & NUMPY)
# -------------------------------------------------------------

# Task A: Load tables into DataFrames
emp_df = pd.read_sql_query("SELECT * FROM employees;", conn)
dept_df = pd.read_sql_query("SELECT * FROM departments;", conn)
proj_df = pd.read_sql_query("SELECT * FROM projects;", conn)

# Task B: Quick inspection
print("\n--- Employee DF Info ---")
emp_df.info()
print("\n--- Summary Stats ---")
print(emp_df.describe())
print("\nShape:", emp_df.shape)
print("\nNull check:\n", emp_df.isnull().sum())

# Task C: NumPy stats
salaries_arr = emp_df['salary'].to_numpy()

sal_mean = np.mean(salaries_arr)
sal_median = np.median(salaries_arr)
sal_std = np.std(salaries_arr)
sal_min = np.min(salaries_arr)
sal_max = np.max(salaries_arr)

# Annual salary calculation
annual_salaries = salaries_arr * 12
emp_df['annual_salary'] = annual_salaries

# Task D: Pandas aggregations
city_avg_salary = emp_df.groupby('city')['salary'].mean()
dept_payroll = emp_df.groupby('dept_id')['salary'].sum()
top_avg_city = city_avg_salary.idxmax()

# Task E: Merge (mirroring Q4 INNER JOIN)
merged_inner = pd.merge(emp_df, dept_df, on='dept_id', how='inner')
merged_check = merged_inner[['name', 'dept_name']]

# -------------------------------------------------------------
# BONUS TASKS
# -------------------------------------------------------------

# 1. City filter helper
def filter_by_city(city_name):
    return emp_df[emp_df['city'].str.lower() == city_name.lower().strip()]

# 2. Custom salary grading
conditions = [
    emp_df['salary'] > 100000,
    (emp_df['salary'] >= 50000) & (emp_df['salary'] <= 100000),
    emp_df['salary'] < 50000
]
grades = ['Grade A', 'Grade B', 'Grade C']
emp_df['salary_grade'] = np.select(conditions, grades, default='N/A')

# 3. Find unassigned staff
unassigned_sql = """
SELECT e.name 
FROM employees e 
LEFT JOIN projects p ON e.emp_id = p.emp_id 
WHERE p.project_id IS NULL;
"""
df_unassigned_sql = pd.read_sql_query(unassigned_sql, conn)

# Unassigned via pandas
unassigned_pd = emp_df[~emp_df['emp_id'].isin(proj_df['emp_id'])][['name']]

# Close connection when finished
conn.close()
```
